```python
import math

# Geometry (mm)
b = 300.0      # beam width
h = 450.0      # beam height
d = 400.0      # effective depth

num_of_rebar = 3
dia_of_rebar = 16.0

# Material properties (MPa = N/mm²)
fc = 28.0      # concrete compressive strength
fy = 420.0     # steel yield strength
Es = 200000.0  # steel modulus of elasticity

# Factored design moment
Mu = 80.0      # kN.m

# Strength reduction factor
phi = 0.9

# Ultimate concrete strain
epsilon_cu = 0.003

# Area of steel reinforcement
As = num_of_rebar * math.pi * dia_of_rebar**2 / 4  # mm²

# ACI 318 beta1 factor
beta1 = max(0.65, 0.85 - 0.05 * max(fc - 28.0, 0.0) / 7.0)

# Equivalent rectangular stress block
a = As * fy / (0.85 * fc * b)   # mm
c = a / beta1                   # mm

# Steel strain
epsilon_s = epsilon_cu * (d - c) / c
epsilon_y = fy / Es

# Nominal moment capacity (N.mm)
Mn = As * fy * (d - a / 2)

# Design strength (N.mm)
Mr = phi * Mn

# Convert to kN.m
Mn_kNm = Mn / 1e6
Mr_kNm = Mr / 1e6

print(f"Area of steel reinforcement As = {As:.1f} mm²")
print(f"Depth of equivalent stress block a = {a:.1f} mm")
print(f"Neutral axis depth c = {c:.1f} mm")
print(f"Rebar strain εs = {epsilon_s:.5f}")
print(f"Steel yield strain εy = {epsilon_y:.5f}")
print(f"Nominal moment capacity Mn = {Mn_kNm:.2f} kN.m")
print(f"Design moment capacity Mr = {Mr_kNm:.2f} kN.m")

if Mr_kNm >= Mu:
    print("PASS : Section is adequate for flexure.")
else:
    print("FAIL : Section is inadequate for flexure.")
```
