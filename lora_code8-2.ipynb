```python
import numpy as np
import matplotlib.pyplot as plt

a=400 #mm, section height
b=300 #mm, section width
c=50 #mm, concrete cover
d=a-c
As1=1017 #mm2, 4phi18, bottom rebar area
As2=452 #mm2, 4phi12, top rebar area

#concrete
fck=25 #MPa
gamma_c=1.5
fcd=0.85*fck/gamma_c #MPa
eps_cc=-0.0020
eps_cu=-0.0035

#steel
fyk=450 #MPa
gamma_s=1.15
fyd=fyk/gamma_s #MPa
Es=200000 #MPa
eps_su=0.01
eps_y=fyd/Es

def rel_c(eps):
    eps = np.asarray(eps, dtype=float)
    
    sig = np.zeros_like(eps)

    parabolic_region = (eps < 0.0) & (eps >= eps_cc)
    plastic_region = (eps < eps_cc) & (eps >= eps_cu)
    invalid_region = eps < eps_cu

    sig[parabolic_region] = (-fcd * (2.0 - eps[parabolic_region] / eps_cc)
                                  * (eps[parabolic_region] / eps_cc))
    
    sig[plastic_region] = -fcd

    if np.any(invalid_region):
        raise ValueError("Beton birim şekil değiştirmesi eps_cu değerinden küçük olamaz.")
    return sig


```
