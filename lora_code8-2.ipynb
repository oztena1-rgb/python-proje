```python
import numpy as np
import matplotlib.pyplot as plt

a=400 #mm, section height
b=300 #mm, section width
c=50 #mm, concrete cover
d=a-c
As1=1017 #mm2, 4phi18, bottom rebar area
As2=452 #mm2, 4phi12, top rebar area

#concrete
fck=25 #MPa
gamma_c=1.5
fcd=0.85*fck/gamma_c #MPa
eps_cc=-0.0020
eps_cu=-0.0035

#steel
fyk=450 #MPa
gamma_s=1.15
fyd=fyk/gamma_s #MPa
Es=200000 #MPa
eps_su=0.01
eps_y=fyd/Es

def rel_c(eps):
    n=len(eps)
    sig=np.zeros(n)

    for i in range(n):

        # Çekme bölgesi
        if eps[i]>=0:
            sig[i]=0
            print(eps[i], "Çekme bölgesi")

        # Parabolik basınç bölgesi
        elif eps[i] >= eps_cc:
            sig[i]=(
                -fcd
                *(2-eps[i]/eps_cc)
                *(eps[i]/eps_cc)
            )
            print(eps[i],"parabolik bölge")

        # Sabit basınç bölgesi
        elif eps[i] >= eps_cu:
            sig[i]=-fcd
            print(eps[i],"sabit basınç bölgesi")

        # Geçersiz değer
        else:
            sig[i] = np.nan
            print(eps[i],"Geçersiz bölge:")

    return sig

def rel_s(eps):
    n=len(eps)
    sig=np.zeros(n)
    for i in range(n):
        # Basınçta akma bölgesi
        if -eps_su <= eps[i] <=-eps_y:
            sig[i]=-fyd

        # Elastik Bölge
        elif -eps_y < eps[i] <= eps_y:
            sig[i]=eps[i]*Es

        # Çekmede Akma Bölgesi 
        elif eps_y < eps[i] <= eps_su:
            sig[i]=fyd
        else:
            print('invalid eps value')
    return(sig)

eps_test =np.array([
0.0,
-0.0005,
-0.0010,
-0.0015,
-0.0020,
-0.0025,
-0.0030,
-0.0035  
])

sig_test = rel_c(eps_test)

print("Beton gerilme-birim şekil değiştirme sonuçları")

for i in range(len(eps_test)):
    print(
        f"eps = {eps_test[i]:6f}, "
        f"sigma = {sig_test[i]:.4f} MPa"
    )
print("eps_cc =", eps_cc)
print("eps_cu=", eps_cu)
print("eps_test=", eps_test)
sig_test = rel_s(eps_test)

print("Birim şekil değiştirme:", eps_test)
print("Çelik gerilmesi:", sig_test)
```
