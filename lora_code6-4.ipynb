```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_excel("beam.csv.xlsx")

fig, ax=plt.subplots()

ax.plot(data["x"], data["M1"], label="M1")
ax.set_xlabel("x")
ax.set_ylabel("bending moment")
ax.set_title("Beam")
ax.legend()
ax.invert_yaxis()
fig.set_size_inches(10, 3)
plt.show()
```
