```python
import pandas as pd
import matplotlib.pyplot as plt

data=pd.read_excel("beam.csv.xlsx")

fig, ax=plt.subplots(figsize=(8,3))
ax.plot(data["x"], data["M1"], label="M1", linewidth=3)
ax.fill_between(data["x"], data["M1"], color="lightblue", alpha=0.5)
ax.invert_yaxis()
ax.legend()
ax.grid(True)
```
