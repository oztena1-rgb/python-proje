```python
import numpy as np

#-------------------------------------------------
# Kesit ve malzeme bilgileri
#-------------------------------------------------

n_step=10000

a = 0.30            # Kesit yüksekliği (m)
c= 0.05             # Donatı ekseninin yüzeyden uzaklığı (m)

As1 = 0.001         # Üst donatı alanı (m^2)
As2 = 0.001         # Alt donatı alanı (m^2)

eps_c = 0.0035        #üst yüz basınç gerinimi

Es = 200e9          # Çelik elastisite modülü (Pa)
fy = 420e6          # Çelik akma dayanımı (Pa)

# Donatıların üst yüzden koordinatları
y_s1 = c
y_s2 = a - c

# -------------------------------------------------
# Çeliğin gerilme-gerinim bağıntısı
# -------------------------------------------------

def steel_stress_strain(eps_s):
    """
    Çelik geriniminden gerilimi hesaplar.
    Gerilme +fy ve -fy değerleri arasında sınırlandırılır.
    """
    sigma = Es * eps_s
    return np.clip(sigma, -fy, fy)

#---------------------------------------------------
# Tarafsız eksen için deneme değerleri
#---------------------------------------------------

yn = np.linspace(0.001, a-0.001, n_step)

#Eğrilik
psi = eps_c / yn

# Donatı gerinimleri
# Pozitif: çekme
# Negatif: basınç

eps_s1 = psi * (y_s1-yn)
eps_s2 = psi * (y_s2-yn)

# Donatı gerinimleri
# Pozitif: çekme
# Negatif: basınç

eps_s1 = psi * (y_s1 - yn)
eps_s2 = psi * (y_s2 - yn)

# Donatı gerilmeleri
sig_s1 = steel_stress_strain(eps_s1)
sig_s2 = steel_stress_strain(eps_s2)

# Donatı kuvvetleri
F_s1 = As1 * sig_s1
F_s2 = As2 * sig_s2

#---------------------------------------------------
# Normal kuvvet dengesi
# Saf eğilmede N=0 olmalıdır
#----------------------------------------------------

Nrd = F_s1 + F_s2

#Nrd değerini sıfıra en çok yaklaştıran indis
index = np.argmin(np.abs(Nrd))

# Dengeyi sağlayan sonuçlar
yn_result =yn[index]
psi_result = psi[index]

eps_s1_result = eps_s1[index]
eps_s2_result = eps_s2[index]

sig_s1_result = sig_s1[index]
sig_s2_result = sig_s2[index]

F_s1_result = F_s1[index]
F_s2_result = F_s2[index]

Nrd_result = Nrd[index]

#---------------------------------------------------
# Kesit orta noktasına göre moment hesabı
#---------------------------------------------------
Mrd_result = (F_s1_result * (y_s1 - a/2) + F_s2_result * (y_s2 - a/2))

#--------------------------------------------------
# Sonuçların yazdırılması
#--------------------------------------------------
print("Tarafsız eksen konumu (yn) = {:.6f} m")
print("Eğrilik (psi)              = {:.6f} 1/m")

print(f"üst donatı gerinimi eps_s1 = {eps_s1_result:.6f}")
print(f"alt donatı gerinimi eps_s2 = {eps_s2_result:.6f}")

print(f"üst donatı gerilimi sig_s1 = {sig_s1_result / 1e6:.2f} MPa")
print(f"alt donatı gerilimi sig_s2 = {sig_s2_result / 1e6:.2f} MPa")

print(f"Normal kuvvet dengesi Nrd = {Nrd_result / 1E3} kN")
print(f"Moment Mrd                = {abs(Mrd_result) / 1E3:.3f} kN.m")
    
```
