```python
import numpy as np
n_step=20

eps = 0.0035
n=1
c=0.05
a=0.30

As1=0.001
As2=0.001

#Gerinim aralığı
yn_sup=-9999
yn_inf=0

yn = np.linspace(0.001, 1.0, n_step)

#Her yn değeri için psi hesabı
#Kesit boyunca gerinimdağılımı
psi = eps ** (1/(4 * yn))

#Gerinimler
eps_s1 = n * psi
eps_s2 = psi * (yn - c)

# Gerilme-gerinim bağıntısı
def rel_s(eps_value):
    """
    Gerinim değerinden gerilme değerini hesaplar.
    Buradaki bağıntı kullanılan malzeme modeline göre
    değiştirilmelidir.
    """
    E_s = 200e9
    return E_s * eps_value

# Gerilmeler
sig_s1=rel_s(eps_s1)
sig_s2=rel_s(eps_s2)

print("psi =", psi)
print("eps_s1 =",eps_s1)
print("eps_s2 =",eps_s2)
print("sig_s1 =",sig_s1)
print("sig_s2 =",sig_s2)

# Normal Kuvvet
Nrd_1 = As1 * sig_s1 + As2 * sig_s2

# Moment
Mrd_1 = As1 * sig_s1 * (a/2-c) - As2 * sig_s2 * (a/2-c)

#Sonuçlar
print("yn =", yn)
print("psi =", psi)
print("eps_s1=", eps_s1)
print("eps_s2=", eps_s2)
print("sig_s1 =", sig_s1)
print("sig_s2 =", sig_s2)
print("sig_s1 =", sig_s1)
print("Nrd_1=", Nrd_1)
print("Mrd_1=", Mrd_1)
```
