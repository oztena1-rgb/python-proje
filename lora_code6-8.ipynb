```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_excel("beam.csv.xlsx")

plt.style.use("seaborn-v0_8-whitegrid")
fig, ax=plt.subplots(figsize=(8,3))
ax.plot(data["x"], data["M1"],label="M1", linewidth=3)
ax.plot(data["x"], data["M2"],label="M2", linewidth=3)

ax.set_xlabel("x(m)")
ax.set_ylabel("Moment")
ax.set_title("Moment Distribution")

ax.axhline(0, color="black", linewidth=1)
ax.invert_yaxis()
ax.grid(True)
ax.legend()

plt.show()
```
