```python
import numpy as np
import sympy as sp
import pandas as pd
sp.init_printing(use_latex="mathjax")

b=250 #mm
B=300 #mm
h=400 #mm
tf1=18 #mm
tf2=15 #mm
tw=12 #mm

hw=h-tf1-tf2
Af1=b*tf1
Af2=B*tf2
Aw=hw*tw
A=Af1+Af2+Aw
A
yG=sp.symbols('yG')
Stot=b*tf1*(yG-tf1/2)-B*tf2*(h-yG-tf2/2)\
            +tw*(yG-tf1)**2/2-tw*(h-yG-tf2)**2/2
yG=sp.solveset(Stot, yG).args[0]
yG
```
