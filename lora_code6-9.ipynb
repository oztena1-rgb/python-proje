```python
import pandas as pd
import matplotlib.pyplot as plt

data =pd.read_excel("beam.csv.xlsx")

fig, ax=plt.subplots(figsize=(8,3))

ax.invert_yaxis()
ax.plot(data["x"], data["M1"], label="M1", linewidth=3)
ax.plot(data["x"], data["M2"], label="M2", linewidth=3)
ax.fill_between(data["x"], data["M1"], data["M2"], color="gray", alpha=0.3, interpolate=True)

ax.set_xlabel("x(m)")
ax.set_ylabel("Moment(kNm)")
ax.set title("MOment Diyagramı")

```
