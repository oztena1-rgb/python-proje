```python
import numpy as np
import matplotlib.pyplot as plt

a=400 #mm, section height
b=300 #mm, section width
c=50 #mm, concrete cover
d=a-c
As1=1017 #mm2, 4phi18, bottom rebar area
As2=452 #mm2, 4phi12, top rebar area

#concrete
fck=25 #MPa
gamma_c=1.5
fcd=0.85*fck/gamma_c #MPa
eps_cc=-0.0020
eps_cu=-0.0035

#steel
fyk=450 #MPa
gamma_s=1.15
fyd=fyk/gamma_s #MPa
Es=200000 #MPa
eps_su=0.01
eps_y=fyd/Es

def rel_c(eps):
    n=len(eps)
    sig=np.zeros(n)

    for i in range(n):

        # Çekme bölgesi
        if eps[i]>=0:
            sig[i]=0

        # Parabolik basınç bölgesi
        elif eps[i] >= eps_cc:
            sig[i]=(
                -fcd
                *(2-eps[i]/eps_cc)
                *(eps[i]/eps_cc)
            )
        # Sabit basınç bölgesi
        elif eps[i] >= eps_cu:
            sig[i]=-fcd

        # Geçersiz değer
        else:
            print("Geçersiz eps değeri:", eps[i])
        sig[i] = np.nan

    return(sig)
eps_test =np.array([
0.0,
-0.0005,
-0.0010,
-0.0015,
-0.0020,
-0.0025,
-0.0030,
-0.0035  
])

sig_test = rel_c(eps_test)

print("Beton gerilme-birim şekil değiştirme sonuçları")

for i in range(len(eps_test)):
    print(
        f"eps = {eps_test[i]:6f}, "
        f"sigma = {sig_test[i]:.4f} MPa"
    )
```
