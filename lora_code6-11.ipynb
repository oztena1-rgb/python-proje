```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_excel("beam.csv.xlsx")

fig, ax = plt.subplots(figsize=(8, 3))
ax.invert_yaxis()
ax.plot(data["x"], data["M1"], label="M1", linewidth=3, alpha=0.3)
ax.plot(data["x"], data["M2"], label="M2", linewidth=3, alpha=0.3)
ax.fill_between(data["x"], data["M1"], color="lightblue", alpha=0.5)
ax.fill_between(data["x"], data["M2"], color="lightblue", alpha=0.5)

ax.plot([0,4,9], [0,0,0] , linewidth=2, color="dimgrey",
        marker=6, markersize=16)
ax.legend()
ax.grid(True)
```
