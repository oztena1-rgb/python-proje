```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_excel("beam.csv.xlsx")
fig, (ax1, ax2)= plt.subplots(nrows=2, ncols=1, sharex=True, figsize=(8,6))
ax1.plot(data["x"], data["M1"], color="blue", linewidth=2, label="M1")
ax1.set_ylabel("bending moment")
ax1.set_title("combo1")
ax1.invert_yaxis()

ax2.plot(data["x"], data["M2"], color="red", linewidth=2, label="M2")
ax2.set_ylabel("bending moment")
ax2.set_title("combo2")
ax2.invert_yaxis()

plt.show()
```
