```python
import pandas as pd
import matplotlib. pyplot as plt
fig, ((ax1, ax2),(ax3,ax4))= plt.subplots(nrows=2, ncols=2, sharex=True)
fig.set_size_inches(8, 4)
data = pd.read_csv("beam_example.csv", skiprows=1)
data = data.dropna(subset=["x"])

fig,((ax1, ax2), (ax3,ax4)) = plt.subplots(
    nrows=2,
    ncols=2,
    sharex=True
)
fig.set_size_inches(10,6)

ax1.plot(data["x"], data["M neg"], linewidth=3)
ax1.set_ylabel("Moment")
ax1.set_title("Negative Moment")
ax1.invert_yaxis()
ax1.grid()

ax2.plot(data["x"], data["T neg"], linewidth=3)
ax2.set_ylabel("kesme kuvveti")
ax2.set_title("Negative Kesme")
ax2.grid()

ax3.plot(data["x"], data["M npos"], linewidth=3)
ax3.set_ylabel("Moment")
ax3.set_title("Positive Moment")
ax3.invert_yaxis()
ax3.grid()

ax4.plot(data["x"], data["T pos"], linewidth=3)
ax4.set_ylabel("kesme kuvveti")
ax4.set_title("Positive Kesme")
ax4.grid()

plt.tight_layout()
plt.show()
```
