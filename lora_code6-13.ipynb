```python
import pandas as pd
import matplotlib. pyplot as plt
fig, ((ax1, ax2),(ax3,ax4))= plt.subplots(nrows=2, ncols=2, sharex=True)
fig.set_size_inches(8, 4)
data= pd.read_csv("beam_example.csv", skiprows=1)
data = data.rename(columns=(
    "M neg":"M1",
    "T neg":"V1"
    "M pos":"M2"
    "T pos":"V2"
))
data = data.dropna(subset=["x"])

ax1.plot(data["x"],data["M1"], label="M1",linewidth=3)
ax1.set_ylabel("bending moment")
ax1.set_title("Combo 1")
ax1.invert_yaxis()

ax2.plot(data["x"],data["V1"], label="V1",linewidth=3)
ax2.set_ylabel("shear force")
ax2.set_title("Combo 1")

ax3.plot(data["x"],data["M2"], label="M2",linewidth=3)
ax3.set_ylabel("bending moment")
ax3.set_title("Combo 2")
ax3.invert_yaxis()

ax4.plot(data["x"],data["V2"], label="V2",linewidth=3)
ax4.set_ylabel("shear force")
ax4.set_title("Combo 2")

plt.tight_layout()
plt.show()
```
