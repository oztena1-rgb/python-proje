```python
import pandas as pd
import matplotlib. pyplot as plt

#Excel dosyasını oku
data = pd.read_excel("beam.csv.xlsx")

#2 satır ve 2 sütundan oluşan grafik alanı
fig,((ax1, ax2), (ax3,ax4)) = plt.subplots(
    nrows=2,
    ncols=2,
    sharex=True
)
fig.set_size_inches(10,6)

# Combo 1 moment diyagramı
ax1.plot(data["x"], data["M1"], linewidth=3)
ax1.set_ylabel("Bending Moment")
ax1.set_title("Combo 1 -  Moment")
ax1.invert_yaxis()
ax1.grid()

# Combo 1 kesme kuvvet diyagramı
ax2.plot(data["x"], data["V1"], linewidth=3)
ax2.set_ylabel("shear force")
ax2.set_title("Combo 1 - Shear")
ax2.grid()

# Combo 2 moment diyagramı
ax3.plot(data["x"], data["M2"], linewidth=3)
ax3.set_xlabel("x")
ax3.set_ylabel("Bending Moment")
ax3.set_title("Combo 2 - Moment")
ax3.invert_yaxis()
ax3.grid()

# Combo 2 kesme kuvvet diyagramı
ax4.plot(data["x"], data["V2"], linewidth=3)
ax4.set_xlabel("x")
ax4.set_ylabel("Shear Force")
ax4.set_title("Combo 2 - Shear")
ax4.grid()

plt.tight_layout()
plt.show()
```
